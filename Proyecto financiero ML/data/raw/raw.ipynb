```python
def obtener_constituyentes_sp500():
    """
    Descarga la lista completa de componentes del S&P 500 desde Wikipedia,
    incluyendo su símbolo bursátil, nombre corporativo y sector GICS.
    
    Retorna:
    --------
    df_sp500 : pd.DataFrame
        DataFrame con columnas: ['Symbol', 'Security', 'GICS Sector', 'GICS Sub-Industry']
    """
    url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
    
    # Headers para emular una solicitud de navegador estándar
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    html = requests.get(url, headers=headers).text
    
    # Lectura de la primera tabla del documento HTML
    tablas = pd.read_html(html)
    df_sp500 = tablas[0]
    
    # Filtrado y renombrado de columnas relevantes
    df_sp500 = df_sp500[['Symbol', 'Security', 'GICS Sector', 'GICS Sub-Industry']].copy()
    
    # Corrección de formato de tickers para compatibilidad con yfinance (ej. BRK.B -> BRK-B)
    df_sp500['Symbol'] = df_sp500['Symbol'].str.replace('.', '-', regex=False)
    
    return df_sp500

# Ejecución y visualización de la distribución sectorial
df_sp500 = obtener_constituyentes_sp500()
print(f"Total de empresas extraídas: {len(df_sp500)}")
print("\nDistribución de empresas por sector GICS:")
print(df_sp500['GICS Sector'].value_counts())
```

### Oservaciones 
Nos damos cuenta rapidamente que el universo de opciones por sector () es muy limitado entonces nos vamos con un universo más grande.

El índice Russell 1000 Index: Captura las 1,000 empresas de mayor capitalización bursátil del mercado estadounidense, representando aproximadamente el 93% del valor de capitalización total de la renta variable de EE. UU.

```python

def obtener_sp_composite_900():
    """
    Función para extraer, limpiar y unificar los componentes del S&P 500 
    y el S&P MidCap 400, generando el universo del S&P Composite 900 
    junto con sus clasificaciones por sector GICS.
    """
    # Encabezados HTTP para evitar bloqueos por parte de los servidores de Wikipedia
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
    
    # 1. Extracción de los componentes del S&P 500 (Large-Cap)
    url_sp500 = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies#External_links"
    df_500 = pd.read_html(requests.get(url_sp500, headers=headers).text)[0]
    df_500 = df_500[['Symbol', 'Security', 'GICS Sector']].copy()
    
    # 2. Extracción de los componentes del S&P MidCap 400 (Mid-Cap)
    url_sp400 = "https://en.wikipedia.org/wiki/List_of_S%26P_400_companies#External_links"
    df_400 = pd.read_html(requests.get(url_sp400, headers=headers).text)[0]
    
    # Estandarización del nombre de la columna de símbolos ('Symbol' o 'Ticker')
    col_simbolo = 'Symbol' if 'Symbol' in df_400.columns else 'Ticker'
    df_400 = df_400[[col_simbolo, 'Security', 'GICS Sector']].copy()
    df_400.rename(columns={col_simbolo: 'Symbol'}, inplace=True)
    
    # 3. Concatenación de ambos índices para construir el S&P Composite 900
    df_composite_900 = pd.concat([df_500, df_400], ignore_index=True)
    
    # 4. Normalización de tickers para compatibilidad con yfinance (ej. BRK.B -> BRK-B)
    df_composite_900['Symbol'] = df_composite_900['Symbol'].str.replace('.', '-', regex=False)
    
    # Eliminación de registros duplicados por seguridad analítica
    df_composite_900 = df_composite_900.drop_duplicates(subset=['Symbol']).reset_index(drop=True)
    
    return df_composite_900

dfEmpresasBase = obtener_sp_composite_900()

sectores_de_interes = [
    'Information Technology', # Crecimiento y liquidez
    'Consumer Staples',       # Consumo Básico
    'Health Care',
    "Consumer Discretionary"            # Salud y Farmacéuticas
]

candidatas_filtradas = dfEmpresasBase[dfEmpresasBase['GICS Sector'].isin(sectores_de_interes)]
lista_tickers = candidatas_filtradas['Symbol'].tolist() # Muestra para auditoría inicial
print("\n")
print(f"Total de empresas encontradas en los sectores objetivo: {len(candidatas_filtradas)}")

# 3. Descarga masiva mediante yfinance (Semana 3)
datos_precios = yf.download(
    tickers=lista_tickers + ['SPY'],
    start="2015-01-01",
    end="2025-08-01",
    auto_adjust=True,
    progress=False
)['Close']
print("\n")
print(f"Series temporales descargadas. Matriz de precios: {datos_precios.shape}")
```
